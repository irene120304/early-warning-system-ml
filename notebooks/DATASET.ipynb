```python
import numpy as np
import pandas as pd

np.random.seed(42)

# =====================================================
# JUMLAH DATA
# =====================================================

N = 500

# =====================================================
# PROPORSI SISWA
# =====================================================

# 35% Rajin & Berprestasi
# 25% Rajin tetapi Kesulitan
# 20% Pintar tetapi Kurang Aktif
# 20% Berisiko

groups = (
    ["excellent"]*175 +
    ["hardworker"]*125 +
    ["smart"]*100 +
    ["risk"]*100
)

np.random.shuffle(groups)

rows = []

# =====================================================
# GENERATE DATA
# =====================================================

for i, g in enumerate(groups):

    if g=="excellent":

        login = np.random.randint(90,160)
        durasi = np.random.normal(35,6)
        tugas = np.random.randint(16,21)
        forum = np.random.randint(8,20)
        quiz_attempt = np.random.randint(5,10)
        terlambat = np.random.randint(0,2)
        quiz = np.random.normal(88,5)
        waktu = np.random.normal(55,8)

    elif g=="hardworker":

        login = np.random.randint(90,160)
        durasi = np.random.normal(42,8)
        tugas = np.random.randint(15,20)
        forum = np.random.randint(10,25)
        quiz_attempt = np.random.randint(6,12)
        terlambat = np.random.randint(1,5)
        quiz = np.random.normal(68,8)
        waktu = np.random.normal(62,10)

    elif g=="smart":

        login = np.random.randint(20,70)
        durasi = np.random.normal(18,6)
        tugas = np.random.randint(10,18)
        forum = np.random.randint(0,8)
        quiz_attempt = np.random.randint(1,5)
        terlambat = np.random.randint(0,3)
        quiz = np.random.normal(90,5)
        waktu = np.random.normal(25,7)

    else:

        login = np.random.randint(5,60)
        durasi = np.random.normal(15,6)
        tugas = np.random.randint(2,12)
        forum = np.random.randint(0,8)
        quiz_attempt = np.random.randint(1,5)
        terlambat = np.random.randint(4,10)
        quiz = np.random.normal(48,12)
        waktu = np.random.normal(18,7)

    # ===============================
    # Tambahkan noise
    # ===============================

    login += np.random.randint(-15,16)
    durasi += np.random.normal(0,4)
    tugas += np.random.randint(-2,3)
    forum += np.random.randint(-3,4)
    quiz_attempt += np.random.randint(-1,2)
    terlambat += np.random.randint(-1,2)
    quiz += np.random.normal(0,6)
    waktu += np.random.normal(0,5)

    # clipping

    login = int(np.clip(login,1,180))
    durasi = round(np.clip(durasi,2,60),2)
    tugas = int(np.clip(tugas,0,20))
    forum = int(np.clip(forum,0,30))
    quiz_attempt = int(np.clip(quiz_attempt,1,15))
    terlambat = int(np.clip(terlambat,0,10))
    quiz = round(np.clip(quiz,0,100),2)
    waktu = round(np.clip(waktu,2,80),2)

    # =====================================================
    # Learning Engagement Score
    # =====================================================

    score = (

        login/180*15 +

        durasi/60*10 +

        tugas/20*20 +

        forum/30*5 +

        quiz_attempt/15*5 +

        quiz/100*35 +

        waktu/80*10 -

        terlambat/10*10 +

        np.random.normal(0,6)

    )

    if score >= 72:
        label = "Tinggi"

    elif score >= 50:
        label = "Sedang"

    else:
        label = "Rendah"

    rows.append([
        f"S{i+1:04}",
        login,
        durasi,
        tugas,
        forum,
        quiz_attempt,
        terlambat,
        quiz,
        waktu,
        label
    ])

# =====================================================
# DATAFRAME
# =====================================================

columns = [

"student_id",
"jumlah_login",
"rata_rata_durasi_sesi",
"jumlah_tugas_dikumpulkan",
"jumlah_postingan_forum",
"jumlah_percobaan_kuis",
"tingkat_keterlambatan_pengumpulan",
"rata_rata_nilai_kuis",
"total_waktu_belajar"

]

df = pd.DataFrame(rows,columns=columns)

# =====================================================
# Tambahkan Outlier (8%)
# =====================================================

outlier = np.random.choice(df.index,
                           size=int(len(df)*0.08),
                           replace=False)

for idx in outlier:

    case = np.random.randint(4)

    if case==0:

        df.loc[idx,"jumlah_login"] = np.random.randint(120,180)
        df.loc[idx,"rata_rata_nilai_kuis"] = np.random.randint(40,65)

    elif case==1:

        df.loc[idx,"jumlah_login"] = np.random.randint(5,25)
        df.loc[idx,"rata_rata_nilai_kuis"] = np.random.randint(90,100)

    elif case==2:

        df.loc[idx,"tingkat_keterlambatan_pengumpulan"] = np.random.randint(7,10)
        df.loc[idx,"rata_rata_nilai_kuis"] = np.random.randint(85,95)

    else:

        df.loc[idx,"jumlah_postingan_forum"] = np.random.randint(20,30)
        df.loc[idx,"rata_rata_nilai_kuis"] = np.random.randint(45,60)

# =====================================================
# SHUFFLE
# =====================================================

df = df.sample(frac=1,random_state=42).reset_index(drop=True)

# =====================================================
# SAVE
# =====================================================

df.to_csv("lms_dataset_realistis.csv",index=False)

print(df.head())

print("\nDistribusi kelas")

print(df["tingkat_performa"].value_counts())

print("\nDataset berhasil dibuat.")
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

N = 500

rows = []

for i in range(N):

    # ------------------------------------------------
    # Kemampuan dasar siswa (tidak disimpan)
    # ------------------------------------------------
    kemampuan = np.random.normal(0.6, 0.2)
    motivasi = np.random.normal(0.6, 0.2)
    disiplin = np.random.normal(0.6, 0.2)

    kemampuan = np.clip(kemampuan, 0, 1)
    motivasi = np.clip(motivasi, 0, 1)
    disiplin = np.clip(disiplin, 0, 1)

    # ------------------------------------------------
    # Login
    # ------------------------------------------------

    login = int(
        15
        + motivasi*80
        + np.random.normal(0,15)
    )

    login = np.clip(login,1,180)

    # ------------------------------------------------
    # Durasi
    # ------------------------------------------------

    durasi = (
        8
        + motivasi*20
        + kemampuan*8
        + np.random.normal(0,4)
    )

    durasi = np.clip(durasi,2,60)

    # ------------------------------------------------
    # Tugas
    # ------------------------------------------------

    tugas = int(

        3
        + disiplin*14
        + np.random.normal(0,2)

    )

    tugas = np.clip(tugas,0,20)

    # ------------------------------------------------
    # Forum
    # ------------------------------------------------

    forum = int(

        motivasi*12
        + np.random.normal(0,4)

    )

    forum = np.clip(forum,0,30)

    # ------------------------------------------------
    # Quiz Attempt
    # ------------------------------------------------

    attempt = int(

        1
        + motivasi*5
        + np.random.normal(0,1)

    )

    attempt = np.clip(attempt,1,10)

    # ------------------------------------------------
    # Terlambat
    # ------------------------------------------------

    terlambat = int(

        (1-disiplin)*6
        + np.random.normal(0,1)

    )

    terlambat = np.clip(terlambat,0,10)

    # ------------------------------------------------
    # Nilai Quiz
    # ------------------------------------------------

    quiz = (

        40
        + kemampuan*45
        + motivasi*8
        - terlambat*2
        + np.random.normal(0,8)

    )

    quiz = np.clip(quiz,0,100)

    # ------------------------------------------------
    # Total Belajar
    # ------------------------------------------------

    belajar = (

        5
        + login*0.25
        + durasi*0.4
        + np.random.normal(0,4)

    )

    belajar = np.clip(belajar,2,80)

    rows.append([

        f"S{i+1:04}",

        login,

        round(durasi,2),

        tugas,

        forum,

        attempt,

        terlambat,

        round(quiz,2),

        round(belajar,2)

    ])

columns = [

    "student_id",
    "jumlah_login",
    "rata_rata_durasi_sesi",
    "jumlah_tugas_dikumpulkan",
    "jumlah_postingan_forum",
    "jumlah_percobaan_kuis",
    "tingkat_keterlambatan_pengumpulan",
    "rata_rata_nilai_kuis",
    "total_waktu_belajar"

]

df = pd.DataFrame(rows, columns=columns)

# ------------------------------------------------
# Tambahkan outlier agar lebih realistis
# ------------------------------------------------

idx = np.random.choice(df.index, 40, replace=False)

for i in idx:

    kasus = np.random.randint(4)

    if kasus == 0:
        # Sering login tetapi nilai kuis rendah
        df.loc[i, "jumlah_login"] = np.random.randint(120,180)
        df.loc[i, "rata_rata_nilai_kuis"] = np.random.randint(45,65)

    elif kasus == 1:
        # Login sedikit tetapi nilai kuis tinggi
        df.loc[i, "jumlah_login"] = np.random.randint(5,20)
        df.loc[i, "rata_rata_nilai_kuis"] = np.random.randint(88,100)

    elif kasus == 2:
        # Sering terlambat tetapi masih cukup baik
        df.loc[i, "tingkat_keterlambatan_pengumpulan"] = np.random.randint(6,10)
        df.loc[i, "rata_rata_nilai_kuis"] = np.random.randint(75,90)

    else:
        # Aktif forum tetapi nilai biasa
        df.loc[i, "jumlah_postingan_forum"] = np.random.randint(20,30)
        df.loc[i, "rata_rata_nilai_kuis"] = np.random.randint(55,70)

# Acak urutan data
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# Simpan
df.to_csv("contoh_upload_dashboard.csv", index=False)

print(df.head())
print("\nJumlah data:", len(df))
print("\nFile berhasil dibuat!")
```
